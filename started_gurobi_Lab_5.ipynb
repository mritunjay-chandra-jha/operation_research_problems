Q1) Problem 1
An advisory firm must allocate ₹50,00,000 across four instruments: Equity (E), Corporate Bonds (B), Government Securities (G), and Money Market (M). Expected annual returns are 14%, 9%, 7%, and 5% respectively.
Constraints imposed by the client’s mandate:
1.	The entire ₹50,00,000 must be invested.
2.	No more than 35% of the total portfolio may be in Equity.
3.	At least twice as much must be held in Government Securities as in Money Market.
4.	The portfolio’s weighted-average return must be at least 9%.
5.	Corporate Bonds cannot exceed Government Securities in amount.



```python
from gurobipy import Model, GRB

m = Model("Investment_Allocation")

E = m.addVar(lb=0, name="Equity")
B = m.addVar(lb=0, name="Corporate_Bonds")
G = m.addVar(lb=0, name="Government_Securities")
M = m.addVar(lb=0, name="Money_Market")

m.setObjective(
    0.14 * E +
    0.09 * B +
    0.07 * G +
    0.05 * M,
    GRB.MAXIMIZE
)

m.addConstr(E + B + G + M == 5000000, name="Total_Investment")

m.addConstr(E <= 0.35 * 5000000, name="Equity_Limit")

m.addConstr(G >= 2 * M, name="Govt_vs_MoneyMarket")

m.addConstr(
    0.14 * E +
    0.09 * B +
    0.07 * G +
    0.05 * M >= 0.09 * 5000000,
    name="Minimum_Return"
)

m.addConstr(B <= G, name="Bonds_vs_Government")

m.optimize()

if m.status == GRB.OPTIMAL:
    print("Optimal solution found")
    print(f"Equity                = ₹{E.X:,.2f}")
    print(f"Corporate Bonds       = ₹{B.X:,.2f}")
    print(f"Government Securities = ₹{G.X:,.2f}")
    print(f"Money Market          = ₹{M.X:,.2f}")
    print(f"Maximum Annual Return = ₹{m.ObjVal:,.2f}")

```

```python
for c in m.getConstrs():
    print(
        f"{c.ConstrName}: "
        f"Slack = {c.Slack:.2f}, "
        f"Shadow Price = {c.Pi:.4f}"
    )
```

```python
from gurobipy import Model, GRB
import pandas as pd

m = Model("Investment_Allocation")

E = m.addVar(lb=0, name="Equity")
B = m.addVar(lb=0, name="Corporate_Bonds")
G = m.addVar(lb=0, name="Government_Securities")
M = m.addVar(lb=0, name="Money_Market")

m.setObjective(
    0.14 * E +
    0.09 * B +
    0.07 * G +
    0.05 * M,
    GRB.MAXIMIZE
)

m.addConstr(E + B + G + M == 5000000, name="Total_Investment")
m.addConstr(E <= 1750000, name="Equity_Limit")
m.addConstr(G >= 2 * M, name="Govt_vs_MoneyMarket")
m.addConstr(
    0.14 * E +
    0.09 * B +
    0.07 * G +
    0.05 * M >= 450000,
    name="Minimum_Return"
)
m.addConstr(B <= G, name="Bonds_vs_Government")

m.optimize()

if m.status == GRB.OPTIMAL:
    print("Optimal solution found")
    print(f"Equity                = ₹{E.X:,.2f}")
    print(f"Corporate Bonds       = ₹{B.X:,.2f}")
    print(f"Government Securities = ₹{G.X:,.2f}")
    print(f"Money Market          = ₹{M.X:,.2f}")
    print(f"Maximum Annual Return = ₹{m.ObjVal:,.2f}")
    print(f"Return Percentage     = {(m.ObjVal / 5000000) * 100:.2f}%")

Report = []

for c in m.getConstrs():
    Report.append({
        "Constraint": c.ConstrName,
        "Slack": c.Slack,
        "Shadow Price": c.Pi,
        "Allowable Min": c.SARHSLow,
        "Allowable Max": c.SARHSUp
    })

df = pd.DataFrame(Report).set_index("Constraint")

print("\nSensitivity Report")
print(df)

return_constraint = m.getConstrByName("Minimum_Return")

print("\nPart D")
print(f"Shadow Price = {return_constraint.Pi:.4f}")
print(f"Allowable Min = {return_constraint.SARHSLow:.2f}")
print(f"Allowable Max = {return_constraint.SARHSUp:.2f}")

old_return = m.ObjVal
old_rhs = return_constraint.RHS
new_rhs = 400000

predicted_return = old_return + return_constraint.Pi * (new_rhs - old_rhs)

print(f"Predicted Return at 8% = ₹{predicted_return:,.2f}")

return_constraint.RHS = new_rhs
m.optimize()

print(f"Actual Return at 8% = ₹{m.ObjVal:,.2f}")

N = m.addVar(lb=0, ub=500000, name="N")

m.setObjective(
    0.14 * E +
    0.09 * B +
    0.07 * G +
    0.05 * M +
    0.20 * N,
    GRB.MAXIMIZE
)

m.remove(m.getConstrByName("Total_Investment"))

m.addConstr(
    E + B + G + M + N == 5000000,
    name="Total_Investment"
)

m.optimize()

if m.status == GRB.OPTIMAL:
    print("\nPart E")
    print(f"Equity                = ₹{E.X:,.2f}")
    print(f"Corporate Bonds       = ₹{B.X:,.2f}")
    print(f"Government Securities = ₹{G.X:,.2f}")
    print(f"Money Market          = ₹{M.X:,.2f}")
    print(f"New Instrument N      = ₹{N.X:,.2f}")
    print(f"Maximum Annual Return = ₹{m.ObjVal:,.2f}")
    print(f"Return Percentage     = {(m.ObjVal / 5000000) * 100:.2f}%")
    print(f"Reduced Cost of N     = {N.RC:.4f}")
```
